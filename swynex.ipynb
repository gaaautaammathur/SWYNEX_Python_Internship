```python
def parse_date(value):
    if pd.isnull(value):                      # nothing to convert if the date is missing
        return pd.NaT                         # NaT = "Not a Time" (missing date)
    for fmt in ["%Y-%m-%d", "%d/%m/%Y", "%m-%d-%Y", "%d-%b-%Y"]:
        try:
            return pd.to_datetime(value, format=fmt)   # works only if the format matches
        except ValueError:
            continue                          # wrong format -> try the next one
    return pd.NaT                             # impossible dates (2025-02-30) end up as missing
```

```python
df["order_date"]    = df["order_date"].apply(parse_date)    # apply() runs the function on every row
df["delivery_date"] = df["delivery_date"].apply(parse_date)
```

```python
# A quantity can't be zero/negative, and 100+ items is unrealistic
df["quantity"] = df["quantity"].mask((df["quantity"] <= 0) | (df["quantity"] > 100))
 
# A price can't be zero/negative, and a very large price is a typing error
df["unit_price"] = df["unit_price"].mask((df["unit_price"] <= 0) | (df["unit_price"] > 200000))
 
# A discount must be between 0% and 100%
df["discount_pct"] = df["discount_pct"].mask((df["discount_pct"] < 0) | (df["discount_pct"] > 100))
 
# A rating must be between 1 and 5
df["rating"] = df["rating"].mask(~df["rating"].between(1, 5))
 
# %%
# An order can't be placed in the future
df["order_date"] = df["order_date"].mask(df["order_date"] > pd.Timestamp.today())
 
# A product can't be delivered before it was ordered
df["delivery_date"] = df["delivery_date"].mask(df["delivery_date"] < df["order_date"])
 
# Only delivered orders can have a rating
df["rating"] = df["rating"].mask(df["order_status"] != "Delivered")
```

```python
df["email"]
```

```python
valid_email = df["email"].str.match(r"^[^@\s]+@[^@\s]+\.[^@\s]+$", na=False)
df["email"] = df["email"].mask(~valid_email)   # invalid emails -> missing
```

```python
df["phone"].unique()
```

```python
phone = df["phone"].astype(str).str.replace(r"\D", "", regex=True)   # \D = anything that is not a digit
phone = phone.str[-10:].where(phone.str.len() >= 10)                 # last 10 digits
df["phone"] = phone.where(phone.str.len() == 10)  
```

```python
before = len(df)                                           # rows before removing duplicates
df = df.drop_duplicates()                                  # removes rows that are exactly the same
df = df.drop_duplicates(subset="order_id", keep="first")   # one order_id = one order, keep the first
print("Rows removed:", before - len(df))
```

```python
product_category = df.groupby("product_name")["category"].agg(lambda s: s.mode()[0])
```

```python
product_category
```

```python
df["unit_price"]
```

```python
product_price    = df.groupby("product_name")["unit_price"].agg(lambda s: s.mode()[0])
```

```python
product_price  
```

```python
df["category"]   = df["category"].fillna(df["product_name"].map(product_category))
df["unit_price"] = df["unit_price"].fillna(df["product_name"].map(product_price))
```

```python
df["category"]
```

```python
df["quantity"]       = df["quantity"].fillna(df["quantity"].median())
```

```python
df["discount_pct"]   = df["discount_pct"].fillna(0) 
```

```python
df["payment_method"] = df["payment_method"].fillna("Unknown") 
```

```python
df["city"]           = df["city"].fillna("Unknown")
df["state"]          = df["state"].fillna("Unknown")
```

```python
df["total_amount"] = (df["quantity"] * df["unit_price"] * (1 - df["discount_pct"] / 100)).round(2)
```

```python
print(df.isnull().sum())
```

```python
print("Duplicate order_ids:", df["order_id"].duplicated().sum())
```

```python
for col in ["customer_name", "email"]:
    lookup = df.dropna(subset=[col]).groupby("customer_id")[col].agg(lambda s: s.mode()[0])
    df[col] = df[col].fillna(df["customer_id"].map(lookup))
```
